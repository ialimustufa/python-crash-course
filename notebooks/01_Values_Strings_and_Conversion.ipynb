## Values and types

| Type | Example | Meaning |
|---|---|---|
| `int` | `7` | whole number |

## Conversion

If a value came from a form or a CSV cell, it is often a **string**, even when it looks like `23`. Convert with `int()`, `float()`, or `str()`.

`int("19.0")` fails because of the decimal point. Use `int(float("19.0"))` when you need a whole number.

Bad conversions raise `ValueError`. Notebook 05 will catch that on purpose.

![A CSV cell is text until you convert it on purpose](https://mermaid.ink/svg/Zmxvd2NoYXJ0IExSCiAgY3N2Q2VsbFsiQ1NWIGNlbGwgMjMiXSAtLT4gYXNUZXh0W3N0cl0KICBhc1RleHQgLS0-IGludEZuWyJpbnQoKSJdCiAgaW50Rm4gLS0-IGFzTnVtYmVyWzIzXQogIGFzVGV4dCAtLT4gZmxvYXRGblsiZmxvYXQoKSJdCiAgZmxvYXRGbiAtLT4gYXNGbG9hdFsiMjMuMCJd)

````python
raw_age = "23"          # a CSV cell or form field, not a number yet
age = int(raw_age)
print(age, type(age))
print(age + 1)

print(float("3.5"))
print(str(40) + " km")
try:
    int("19.0")
except ValueError as exc:
    print("int on 19.0 failed:", exc)
print("via float:", int(float("19.0")))
````

## Strings: index, slice, methods, f-strings

Indexing starts at `0`. Slices are `[start:stop]` and **exclude** `stop`. Negative indexes count from the end.

An f-string puts values into text: `f"{name} logged {trips} trips."`

## Practice 01.1


Convert 22°C to Fahrenheit: `f = c * 9 / 5 + 32`.


````python
# Practice 01.1
celsius = 22
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)

# Check 01.1
assert fahrenheit == 71.6
print("Practice 01.1 passed.")
````

<summary>Hint 01.1</summary>

Use `*` and `/` before `+`. Parentheses are optional here because of operator order.

</details>


## Practice 01.2


From `full = "Ada Lovelace"`, get the first name and last name with slicing and `split`.


````python
# Practice 01.2
full = "Ada Lovelace"
first_name = full.split()[0]
last_name = full.split()[1]
print(first_name, last_name)

# Check 01.2
assert first_name == "Ada"
assert last_name == "Lovelace"
print("Practice 01.2 passed.")
````

<details>
<summary>Hint 01.2</summary>

`full.split()` returns `['Ada', 'Lovelace']`. Index `0` and `1`.

</details>


## Practice 01.3


Convert the string `"19.0"` to an integer number of years. Truncate toward zero via `int(float(...))`.


````python
# Practice 01.3
raw = "19.0"
years = int(float(raw))
print(years)

# Check 01.3
assert years == 19 and isinstance(years, int)
print("Practice 01.3 passed.")
````

<details>
<summary>Hint 01.3</summary>

`int("19.0")` fails. Convert to `float` first, then to `int`.

</details>


## Common mistakes

- `+` on strings concatenates: `"3" + "4"` is `"34"`.
- `/` always returns a `float` in Python 3: `4 / 2` is `2.0`. Use `//` for integer division.
- Slices never include the end index: `"Python"[0:3]` is `Pyt`.


## Recap


Types decide what operations are legal. Convert on purpose. Strings are sequences you can index, slice, and format with f-strings.
