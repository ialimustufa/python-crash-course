## Values and types

| Type | Example | Meaning |
|---|---|---|
| `int` | `7` | whole number |
| `float` | `7.5` | number with a fraction |
| `bool` | `True` / `False` | yes or no |
| `None` | `None` | missing / no value |
| `str` | `"ride"` | text |

`type(x)` reports the type. Python will not add a number to text until you convert one of them.


````python
trips = 7
distance_km = 12.5
active = True
note = None
label = "commute"

print(type(trips), type(distance_km), type(active), type(note), type(label))
print(trips * distance_km)
print(3 == 3, 3 != 4, 3 < 4)
print(True and False, True or False, not False)
````

## Conversion

`input()` in Colab always gives a **string**, even if the person typed `23`. Convert with `int()`, `float()`, or `str()`.

Bad conversions raise `ValueError`. Later you will catch that. For now, convert values you already know are numeric.


````python
raw_age = "23"          # what input() would have returned
age = int(raw_age)
print(age, type(age))
print(age + 1)

print(float("3.5"))
print(str(40) + " km")
````

## Strings: index, slice, methods

Indexing starts at `0`. Slices are `[start:stop]` and **exclude** `stop`. Negative indexes count from the end.


````python
city = "Hyderabad"
print(city[0], city[-1])
print(city[0:5])
print(city[::-1])
print(city.lower(), city.upper())
print(city.replace("bad", "baad"))
print("Hyder" in city)
print(len(city))

name = "ana"
print(f"{name.title()} logged {trips} trips.")
````

## Practice 01.1


Convert 22°C to Fahrenheit: `f = c * 9 / 5 + 32`.


````python
# Practice 01.1
celsius = 22
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)

# Check 01.1
assert fahrenheit == 71.6
print("Practice 01.1 passed.")
````

<details>
<summary>Hint 01.1</summary>

Use `*` and `/` before `+`. Parentheses are optional here because of operator order.

</details>


## Practice 01.2


From `full = "Ada Lovelace"`, get the first name and last name with slicing and `split`.


````python
# Practice 01.2
full = "Ada Lovelace"
first_name = full.split()[0]
last_name = full.split()[1]
print(first_name, last_name)

# Check 01.2
assert first_name == "Ada"
assert last_name == "Lovelace"
print("Practice 01.2 passed.")
````

<details>
<summary>Hint 01.2</summary>

`full.split()` returns `['Ada', 'Lovelace']`. Index `0` and `1`.

</details>


## Practice 01.3


Convert the string `"19.0"` to an integer number of years. Truncate toward zero via `int(float(...))`.


````python
# Practice 01.3
raw = "19.0"
years = int(float(raw))
print(years)

# Check 01.3
assert years == 19 and isinstance(years, int)
print("Practice 01.3 passed.")
````

<details>
<summary>Hint 01.3</summary>

`int("19.0")` fails. Convert to `float` first, then to `int`.

</details>


## Common mistakes

- `+` on strings concatenates: `"3" + "4"` is `"34"`.
- `/` always returns a `float` in Python 3: `4 / 2` is `2.0`. Use `//` for integer division.
- Slices never include the end index: `"Python"[0:3]` is `Pyt`.


## Recap


Types decide what operations are legal. Convert on purpose. Strings are sequences you can index, slice, and format.
