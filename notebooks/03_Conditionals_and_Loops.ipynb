## Decisions

## Loops

- `for item in sequence:` visits each value.
- `range(n)` is `0 .. n-1`.
- `enumerate(sequence)` gives `(index, value)`.
- `zip(a, b)` walks two sequences together.
- `while` repeats until the condition becomes false. Update the variable, or it never ends.

Nested loops are just loops inside loops, not a third kind of loop.


````python
for n in range(3):
    print("range", n)

names = ["Ada", "Grace"]
for index, name in enumerate(names):
    print(index, name)

hours = [1, 2]
km = [4.0, 9.5]
for h, d in zip(hours, km):
    print(h, "h,", d, "km")

total = 0
for h in hours:
    total += h
print("total hours", total)
````

## Practice 03.1


Classify `score` as `"fail"` (< 40), `"pass"` (40–74), or `"distinction"` (75+).


````python
# Practice 03.1
score = 81
if score < 40:
    grade = "fail"
elif score < 75:
    grade = "pass"
else:
    grade = "distinction"
print(grade)

# Check 03.1
assert grade == "distinction"
print("Practice 03.1 passed.")
````

<details>
<summary>Hint 03.1</summary>

Use `if`, `elif`, `else`. 81 is not less than 75, so it should take the last branch.

</details>


## Practice 03.2


Count how many trips in `km_list` are at least 5 km.


````python
# Practice 03.2
km_list = [1.2, 8.0, 5.0, 3.4, 12.1]
long_trips = 0
for km in km_list:
    if km >= 5:
        long_trips += 1
print(long_trips)

# Check 03.2
assert long_trips == 3
print("Practice 03.2 passed.")
````

<details>
<summary>Hint 03.2</summary>

Start at `0`. Add `1` when `km >= 5`.

</details>


## Practice 03.3


Use `zip` to build `"Mumbai:2.0"` style strings.


````python
# Practice 03.3
cities = ["Mumbai", "Pune"]
hours = [2.0, 1.5]
labels = []
for city, hour in zip(cities, hours):
    labels.append(f"{city}:{hour}")
print(labels)

# Check 03.3
assert labels == ["Mumbai:2.0", "Pune:1.5"]
print("Practice 03.3 passed.")
````

<details>
<summary>Hint 03.3</summary>

`for city, hour in zip(cities, hours):` gives one pair per step.

</details>


## Practice 03.4


Flatten `[[1, 2], [3, 4]]` into `[1, 2, 3, 4]` with a nested loop.


````python
# Practice 03.4
matrix = [[1, 2], [3, 4]]
flat = []
for row in matrix:
    for value in row:
        flat.append(value)
print(flat)

# Check 03.4
assert flat == [1, 2, 3, 4]
print("Practice 03.4 passed.")
````

<details>
<summary>Hint 03.4</summary>

Outer loop visits each inner list. Inner loop visits each number.

</details>


## Common mistakes

- Writing `if score > 40` and `if score > 75` as two separate `if`s when you needed `elif`.
- Forgetting to update the `while` counter.
- Using `range(len(xs))` when `for x in xs` would be clearer.
